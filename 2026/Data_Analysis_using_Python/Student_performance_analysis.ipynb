The Ordinary Least Squares (OLS) regression model estimated in this study is specified as:
</p>

<p style="
    text-align: center;
    font-size: 16px;
    font-weight: 500;
    color: #00897B;
    font-family: 'Poppins', 'Segoe UI', sans-serif;
">
Performance Index<sub>i</sub> = β₀ + β₁(Hours Studied<sub>i</sub>) + β₂(Previous Scores<sub>i</sub>) + β₃(Sleep Hours<sub>i</sub>) + β₄(Sample Papers Practiced<sub>i</sub>) + β₅(Extracurricular Activities<sub>i</sub>) + ε<sub>i</sub>
</p>

<p style="
    color: #263238;
    font-size: 14px;
    line-height: 1.8;
    font-family: 'Poppins', 'Segoe UI', sans-serif;
">
Where:
</p>

<ul style="
    color: #263238;
    font-size: 14px;
    line-height: 1.8;
    font-family: 'Poppins', 'Segoe UI', sans-serif;
">

<li><strong>β₀</strong> represents the intercept term.</li>

<li><strong>β₁ – β₅</strong> measure the marginal effect of each independent variable on student performance.</li>

<li><strong>ε<sub>i</sub></strong> represents the error term capturing unobserved factors.</li>

</ul>

DEFINING THE VARIABLES:
</h2>

```python
# Define the dependent variable (target variable)
import statsmodels.api as sm

Y = df["Performance Index"]

# Select independent variables (features) for regression analysis
X = df[["Hours Studied",
        "Previous Scores",
        "Sleep Hours",
        "Sample Question Papers Practiced",
        "Extracurricular Activities"]]

# Add a constant term to include the intercept in the regression model
X = sm.add_constant(X)
```

------------------------------------------------------------------------------------------------------------------------

    background-color: #E8EAF6;

    letter-spacing: 0.5px;
">
TRAINING THE MODEL:
</h2>

</div>

```python
# Split the dataset into training (75%) and testing (25%) sets for model validation
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.25, random_state=42
)
```

------------------------------------------------------------------------------------------------------------------------

<div style="
    background-color: #E8EAF6;

<h2 style="

    letter-spacing: 0.5px;
">
RUNNING REGRESSION (OLS) - TRAINING DATA:
</h2>

</div>

```python
# Fit an Ordinary Least Squares (OLS) regression model using the training data
model = sm.OLS(Y_train, X_train).fit()

# Display the regression summary including coefficients, R-squared, and statistical tests
print(model.summary())
```
