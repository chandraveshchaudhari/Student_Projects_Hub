## <span style="color: #1e40af;">Importing Required Libraries</span>

<span style="color: #6b7280;">This section imports the necessary Python libraries for data analysis, visualization, and regression modeling. Pandas and NumPy are used for data handling, Matplotlib and Seaborn for visualization, and Scikit-learn for machine learning and regression analysis.</span>

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Machine learning libraries
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

# Statistical libraries
import statsmodels.api as sm

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Plot settings
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize'] = 18
plt.rcParams['axes.labelsize'] = 14
plt.rcParams['xtick.labelsize'] = 12
plt.rcParams['ytick.labelsize'] = 12
plt.rcParams['legend.fontsize'] = 12
```

## <span style="color: #1e40af;">Data Loading</span>

<span style="color: #6b7280;">The cleaned housing dataset is imported using Pandas. The dataset contains structural and locational characteristics of houses along with their market prices.</span>

```python
# Load dataset
df = pd.read_csv('cleaned_housing_dataset.csv')

# Preview data
print("Dataset Shape:", df.shape)
print(df.head())
```

## <span style="color: #1e40af;">Initial Data Exploration</span>

<span style="color: #6b7280;">Basic exploratory analysis is performed to understand the dataset structure, variable types, and statistical properties. This step helps identify potential issues such as missing values or incorrect data types.</span>

```python
# Dataset information
print("\nDataset Information:")
df.info()
print("\nDataset Shape:", df.shape)

# Summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Missing values check
print("\nMissing Values:")
print(df.isnull().sum())
```

### <span style="color: #374151; font-weight: bold;">Feature Refinement</span>

<span style="color: #6b7280;">Certain variables such as construction year and renovation year are removed because derived variables (house age and renovation indicator) already capture the relevant information more effectively.</span>

```python
# Drop redundant columns
df = df.drop(columns=['Built Year', 'Renovation Year'], errors='ignore')

print(df.head())
```

### <span style="color: #374151; font-weight: bold;">Distribution of Housing Prices</span>

<span style="color: #6b7280;">Understanding the distribution of the dependent variable is essential before modeling. The distribution helps identify skewness and potential need for transformation.</span>

```python
# Price distribution
sns.histplot(df['Price'], kde=True)
plt.title("Distribution of Housing Prices")
plt.show()
```

<span style="color: #6b7280;">The distribution of housing prices exhibits a strong positive skew, with most properties concentrated in the lower to middle price ranges and a small number of high-value properties forming a long right tail. Such skewness is typical in real estate markets due to heterogeneous property characteristics and location premiums. To improve regression performance and stabilize variance, a logarithmic transformation of the price variable may be considered.</span>

```python
df['Log_Price'] = np.log(df['Price'])
```

```python
sns.histplot(df['Log_Price'], kde=True)
plt.title("Distribution of Log Housing Prices")
plt.show()
```

<span style="color: #6b7280;">Correlation analysis is performed to examine the strength and direction of relationships between housing characteristics and price. A correlation matrix helps identify which variables are strongly associated with housing prices and also reveals potential multicollinearity among predictor variables. Understanding these relationships is important before building regression models.</span>

## <span style="color: #1e40af;">Regression Modeling</span>

<span style="color: #6b7280;">Multiple linear regression is applied to quantify the relationship between housing prices and various structural and locational characteristics. Since housing prices exhibit positive skewness, a logarithmic transformation of the price variable is used to improve model performance and satisfy regression assumptions. The transformed variable (Log_Price) is used as the dependent variable in the model.</span>

### <span style="color: #374151; font-weight: bold;">Defining Predictor Variables and Target Variable</span>

<span style="color: #6b7280;">The predictor variables (X) include structural and locational characteristics of houses, while the dependent variable (y) is the logarithm of housing price (Log_Price). The original price variable is excluded from the predictors to avoid redundancy, as the transformed variable is used for modeling.</span>

```python
# Define predictors (independent variables) and target (dependent variable)

X = df.drop(['Price', 'Log_Price'], axis=1)
y = df['Log_Price']

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

### <span style="color: #374151; font-weight: bold;">Train–Test Split</span>

<span style="color: #6b7280;">The dataset is divided into training and testing sets to evaluate the performance of the regression model on unseen data. The training set is used to estimate the model parameters, while the testing set is used to assess prediction accuracy. In this study, approximately 75% of the data is allocated for training and 25% for testing.</span>

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,   # 25% test, 75% train
    random_state=42
)

print("Training set size:", X_train.shape)
print("Testing set size:", X_test.shape)
```

### <span style="color: #374151; font-weight: bold;">Feature Scaling</span>

<span style="color: #6b7280;">Feature scaling is applied to normalize the predictor variables so that variables with different magnitudes contribute proportionately to the regression model. Standardization is performed using the StandardScaler, which transforms the data to have a mean of zero and a standard deviation of one. Scaling improves numerical stability and model performance, particularly when variables are measured in different units.</span>

```python
from sklearn.preprocessing import StandardScaler

# Initialize scaler
scaler = StandardScaler()

# Fit scaler on training data and transform
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using same scaler
X_test_scaled = scaler.transform(X_test)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

### <span style="color: #374151; font-weight: bold;">Linear Regression Model Training</span>

<span style="color: #6b7280;">A multiple linear regression model is trained using the standardized predictor variables to estimate the relationship between housing characteristics and the logarithm of housing prices. The model learns the contribution of each predictor variable in explaining variations in housing prices.</span>

```python
from sklearn.linear_model import LinearRegression

# Initialize model
model = LinearRegression()

# Train the model
model.fit(X_train_scaled, y_train)
```

### <span style="color: #374151; font-weight: bold;">Model Predictions</span>

<span style="color: #6b7280;">After training the regression model, predictions are generated for the testing dataset. These predicted values are compared with the actual values to evaluate the performance and accuracy of the model.</span>

```python
# Predict on test data
y_pred = model.predict(X_test_scaled)

# Display first few predictions
print("First 5 Predictions:", y_pred[:5])
```

### <span style="color: #374151; font-weight: bold;">Model Evaluation</span>

<span style="color: #6b7280;">The performance of the regression model is evaluated using Mean Squared Error (MSE) and the coefficient of determination (R² score). The R² score indicates the proportion of variation in housing prices explained by the predictor variables, while MSE measures the average squared difference between actual and predicted values.</span>

```python
from sklearn.metrics import mean_squared_error, r2_score

# Calculate evaluation metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

<span style="color: #6b7280;">The regression model demonstrates an R² value of approximately 0.775, indicating that about 77.5% of the variation in housing prices is explained by the predictor variables included in the model. This suggests that structural and locational characteristics of houses have substantial explanatory power in determining housing prices. The relatively low Mean Squared Error further indicates that the model provides reasonably accurate predictions on unseen data.</span>

### <span style="color: #374151; font-weight: bold;">Regression Coefficient Analysis</span>

<span style="color: #6b7280;">The regression coefficients indicate the direction and relative magnitude of influence of each predictor variable on housing prices. Since the dependent variable is logarithmic (Log_Price), the coefficients can be interpreted approximately as percentage changes in housing prices associated with a one-unit change in the predictor variable, holding other factors constant.</span>

```python
# Create coefficient table
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

# Sort by importance
coefficients_sorted = coefficients.sort_values(by='Coefficient', ascending=False)

coefficients_sorted
```

<span style="color: #6b7280;">The regression coefficient results indicate that housing quality and location are the most significant determinants of housing prices. The grade of the house shows the strongest positive influence, suggesting that higher construction quality substantially increases property value. Geographic location, captured by latitude, also exhibits a strong positive effect, highlighting the importance of neighborhood characteristics in housing markets.</span>

<span style="color: #6b7280;">Structural variables such as living area and house size contribute positively to price, consistent with economic expectations that larger properties command higher market values. Amenities such as number of views, number of bathrooms, and waterfront presence also show positive relationships with price.</span>

<span style="color: #6b7280;">Interestingly, house age demonstrates a positive coefficient after controlling for other factors, which may reflect the influence of renovation, location advantages, or premium older neighborhoods. Some variables, such as number of bedrooms and proximity-related measures, show relatively smaller effects, possibly due to overlap with other housing characteristics included in the model.</span>

<span style="color: #6b7280;">Overall, the findings support the hedonic pricing framework, where both structural and locational characteristics jointly determine housing prices.</span>

```python
plt.figure(figsize=(8,6))

plt.scatter(y_test, y_pred, alpha=0.5)

plt.xlabel("Actual Log Prices")
plt.ylabel("Predicted Log Prices")

plt.title("Actual vs Predicted Log Housing Prices")

plt.show()
```

### <span style="color: #374151; font-weight: bold;">Actual vs Predicted Values</span>

<span style="color: #6b7280;">The scatter plot compares the actual log housing prices (X-axis) with the predicted values generated by the regression model (Y-axis). The close alignment of observations along the upward diagonal direction indicates strong agreement between observed and predicted values, suggesting that the model captures the relationship between housing characteristics and prices effectively.</span>

<span style="color: #6b7280;">A slightly larger dispersion is observed at higher price levels, which may reflect greater variability in luxury housing markets where additional unobserved factors influence pricing. Overall, the plot confirms that the regression model provides reasonably accurate and reliable predictions.</span>

## <span style="color: #1e40af;">Statistical Validation of the Regression Model</span>

<span style="color: #6b7280;">To evaluate the statistical reliability of the regression model, an Ordinary Least Squares (OLS) regression is performed using the statsmodels library. This approach provides statistical measures such as p-values, t-statistics, confidence intervals, and the F-statistic, which help assess the significance of individual predictors and the overall model fit.</span>
<span style="color: #6b7280;">The dependent variable remains the logarithm of housing prices (Log_Price), consistent with the previous regression analysis.</span>

```python
# Define X and y again (unscaled)
X_sm = df.drop(['Price', 'Log_Price'], axis=1)
y_sm = df['Log_Price']

# Add constant (intercept)
X_sm = sm.add_constant(X_sm)

# Fit OLS model
ols_model = sm.OLS(y_sm, X_sm).fit()

# Display summary
print(ols_model.summary())
```

### <span style="color: #374151; font-weight: bold;">Statistical Validation Results</span>

<span style="color: #6b7280;">The OLS regression results indicate that the overall model is highly statistically significant, as reflected by the large F-statistic and its associated p-value (Prob < 0.001). The R² value of approximately 0.773 suggests that about 77.3% of the variation in housing prices is explained by the predictor variables included in the model. The adjusted R² value is similar, indicating that the model does not suffer from overfitting and that the selected variables contribute meaningfully to price determination.</span>

<span style="color: #6b7280;">Most structural and locational variables exhibit statistically significant coefficients, confirming their importance in explaining housing price variations. Variables related to housing quality, size, and amenities, such as grade of the house, living area, number of bathrooms, waterfront presence, and number of views, show strong positive relationships with prices. These findings are consistent with the hedonic pricing framework in economic theory.</span>

## <span style="color: #1e40af;">Residual Analysis</span>

<span style="color: #6b7280;">Residual analysis is performed to evaluate the assumptions of linear regression. Residuals represent the difference between the actual and predicted values of the dependent variable. A well-fitted regression model should produce residuals that are randomly distributed around zero without any systematic pattern.</span>

<span style="color: #6b7280;">The residual plot helps assess:</span>
- <span style="color: #6b7280;">Linearity of the relationship</span>
- <span style="color: #6b7280;">Constant variance (homoscedasticity)</span>
- <span style="color: #6b7280;">Presence of outliers</span>
- <span style="color: #6b7280;">Model adequacy</span>

```python
# Calculate residuals
residuals = y_test - y_pred

# Residual Plot
plt.figure(figsize=(8,5))
plt.scatter(y_pred, residuals, alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')

plt.xlabel("Predicted Log Prices")
plt.ylabel("Residuals")
plt.title("Residual Plot")

plt.show()
```

The absence of systematic structure suggests that the assumptions of linearity and homoscedasticity are reasonably satisfied.</span>

<span style="color: #6b7280;">The distribution of residuals appears approximately normal, further supporting the reliability of the regression estimates.</span>

## <span style="color: #1e40af;">Multicollinearity Test using Variance Inflation Factor (VIF)</span>

<span style="color: #6b7280;">Multicollinearity occurs when independent variables are highly correlated with each other, which can affect the stability of regression coefficients. The Variance Inflation Factor (VIF) is used to detect this issue, where values above 5–10 indicate high multicollinearity.</span>

<span style="color: #6b7280;">The VIF results show that most variables fall within acceptable ranges, suggesting limited multicollinearity. However, size-related variables such as living area, house area (excluding basement), and basement area exhibit very high or infinite VIF values due to overlapping measurements of housing size. This is expected and does not substantially affect the predictive performance of the model, although it may influence coefficient interpretation.</span>

<span style="color: #6b7280;">Overall, multicollinearity is primarily confined to structurally related variables and does not invalidate the regression analysis.</span>
